```python
import librosa
import numpy as np
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import StandardScaler

# labels = ['happy', 'sad', 'angry', 'neutral']

train_files = ["happy1.wav", "sad1.wav", "angry1.wav", "neutral1.wav"]
train_labels = ["happy", "sad", "angry", "neutral"]

def extract_features(file_path):
    y, sr = librosa.load(file_path, sr=None)
    
    rms = librosa.feature.rms(y=y)
    zcr = librosa.feature.zero_crossing_rate(y)
    mfcc = librosa.feature.mfcc(y=y, sr=sr, n_mfcc=13)
    
    features = [
        np.mean(rms), np.std(rms),
        np.mean(zcr), np.std(zcr)
    ]
    
    for i in range(13):
        features.append(np.mean(mfcc[i]))
        features.append(np.std(mfcc[i]))
    
    return np.array(features)

X_train = np.array([extract_features(f) for f in train_files])
y_train = np.array(train_labels)

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)

model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X_train_scaled, y_train)

test_file = "sad.wav"  
X_test = extract_features(test_file).reshape(1, -1)
X_test_scaled = scaler.transform(X_test)

prediction = model.predict(X_test_scaled)
print("Predition:", prediction[0])
```
